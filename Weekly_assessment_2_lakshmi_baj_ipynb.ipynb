
SECTION A: INTERVIEW THEORY
Q1: Series vs DataFrame Difference & Method Applicability

A Series is a 1-dimensional labeled array holding a single sequence of values, whereas a DataFrame is a 2-dimensional tabular structure with labeled rows and columns.
 * Works on both: .head() or .isnull()
 * Works only on DataFrame: .drop(columns=[...]) or .info()
Q2: Purpose of inplace=True & Why It Is Discouraged

inplace=True modifies the underlying pandas object directly in memory without returning a copy. The pandas core team discourages its use because it rarely saves memory, breaks method chaining, and frequently triggers SettingWithCopyWarning when working with filtered DataFrames.
 * Avoid when: Chaining operations on filtered data subsets, e.g., df_filtered.drop(columns=['a'], inplace=True).

Q3: Difference Between .fillna() and .dropna()

.fillna() replaces missing (NaN) values with a specified value or statistic, while .dropna() deletes the rows or columns containing missing values altogether.
 * Use .fillna(): When missing values represent a known baseline (e.g., imputing missing investor counts as 0).
 * Use .dropna(): When missing values severely impact core analytical metrics and imputation would introduce bias.

Q4: Difference Between pd.to_numeric(col, errors='coerce') and pd.to_numeric(col)

pd.to_numeric(col, errors='coerce') turns unparseable non-numeric strings into NaN, while plain pd.to_numeric(col) throws an error and halts execution if any non-numeric string is found.
 * Real-world usage: Essential when handling dirty real-world datasets containing stray text entries like 'Undisclosed', 'NA', or 'N/A'.

Q5: .loc vs .iloc Indexing

.loc relies on label-based indexing, whereas .iloc relies on integer position-based indexing.
 * Example: df.loc[0] looks for the row explicitly labeled with index label 0. df.iloc[0] retrieves the first physical row in memory regardless of its label.
 * Difference case: In a sorted or filtered DataFrame, index labels stay with their original rows. df_sorted.loc[0] searches for label 0, while df_sorted.iloc[0] grabs the top row in the new sorted order.

Q6: .map() vs .apply() vs .replace()

 * .map(): Maps values in a Series using a dictionary mapping or 1-to-1 callable function.
 * .apply(): Applies an arbitrary function across elements of a Series or along DataFrame axes.
 * .replace(): Swaps specific target values in a Series/DataFrame without needing full dictionary coverage.
 * Usage: Use .map() for key-value dictionary mappings, .apply() for complex multi-variable custom logic, and .replace() for targeted value string cleanups.

Q7: What Triggers SettingWithCopyWarning & How to Fix It

SettingWithCopyWarning occurs when pandas cannot determine if an operation modifies a slice/copy or the original DataFrame during chained assignment.
 * Triggers warning: df_sub = df[df['amount'] > 100]; df_sub['amount'] = 0
 * Fix: Explicitly create an independent copy: df_sub = df[df['amount'] > 100].copy()

Q8: pd.cut() vs pd.qcut()

pd.cut() divides data into equal-width ranges (fixed interval sizes), whereas pd.qcut() divides data into equal-frequency bins (quantiles containing an equal count of records).
 * For income labeling: pd.qcut() is preferred because financial data is heavily right-skewed; pd.qcut() creates balanced distribution buckets (e.g., Low 33%, Mid 33%, High 33%).
